## Question 1
**Create an object named student with the following properties: ● name: "Rahul" ● age: 21 ● courses: an array containing "Math", "Physics", and "Chemistry" Then, perform the following operations: 1. Add a new property grade with the value "A".  2. Update the age to 22. 3. Delete the courses property.**

### Answer

JavaScript Code
```js
let student = {
    name: "Rahul",
    age: 21,
    courses: ["Math", "Physics", "Chemistry"]
};

console.log("Original Object:");

console.log(student);

// Add a new property
student.grade = "A";

// Update age
student.age = 22;

// Delete courses property
delete student.courses;

console.log("Updated Object:");
console.log(student);
Output
Original Object:
{
  name: "Rahul",
  age: 21,
  courses: ["Math", "Physics", "Chemistry"]
}

Updated Object:
{
  name: "Rahul",
  age: 22,
  grade: "A"
}

```
**Explanation**

- student object is created with three properties.
- student.grade = "A" adds a new property named grade.
- student.age = 22 updates the value of age.
- delete student.courses removes the courses property from the object.




## Question 2
**Write a function checkEligibility that takes an age as a parameter
and returns:
● "Eligible" if age is 18 or above
● "Not Eligible" if age is below 18
Use an if-else statement.**

### Answer
JavaScript Code
```js
function checkEligibility(age) {

    if (age >= 18) {
        return "Eligible";
    } else {
        return "Not Eligible";
    }

}

console.log(checkEligibility(20));
console.log(checkEligibility(15));
```

**Output**

Eligible

Not Eligible

**Explanation**

- The function accepts age as a parameter.
- If age is 18 or greater, it returns Eligible.
- Otherwise, it returns Not Eligible.

**Applications**
- Voting eligibility
- Driving license verification
- College admissions
- Job applications

## Question 3
**Write a function getDayName that takes a number (1-7) and returns the
corresponding day of the week. Use a switch statement.**

### Answer
**JavaScript Code**
```js
function getDayName(day) {

    switch(day){

        case 1:
            return "Monday";

        case 2:
            return "Tuesday";

        case 3:
            return "Wednesday";

        case 4:
            return "Thursday";

        case 5:
            return "Friday";

        case 6:
            return "Saturday";

        case 7:
            return "Sunday";

        default:
            return "Invalid Day";
    }

}

console.log(getDayName(1));
console.log(getDayName(5));
console.log(getDayName(8));
```
**Output**

Monday

Friday

Invalid Day

**Explanation**

- The function receives a number from 1 to 7.
- Each case represents a day of the week.
- The default case executes if the input is outside the range.


## Question 4
**Rewrite the checkEligibility function from Question 2 using the
ternary operator.**
### Answer
The ternary operator (? :) is a shorthand way of writing an if-else statement. It evaluates a condition and returns one value if the condition is true and another value if it is false.

**JavaScript Code**
```js
function checkEligibility(age){

    return (age >= 18) ? "Eligible" : "Not Eligible";

}
```
console.log(checkEligibility(21));
console.log(checkEligibility(16));

**Output**

Eligible

Not Eligible

**Explanation**

The syntax of the ternary operator is:

condition ? value_if_true : value_if_false;

**Here:**

If age >= 18 is true → returns Eligible
Otherwise → returns Not Eligible

## Question 5
**Given an array numbers = [1, 2, 3, 4, 5], perform the following: 1. Use a for loop to print each number.
2. Use a while loop to print each number.
3. Use a for-in loop to print each index.
4. Use a for-of loop to print each number.**

### Answer
**JavaScript Code**
```js
const numbers = [1, 2, 3, 4, 5];

// For Loop
console.log("For Loop:");
for (let i = 0; i < numbers.length; i++) {
    console.log(numbers[i]);
}

// While Loop
console.log("While Loop:");
let i = 0;
while (i < numbers.length) {
    console.log(numbers[i]);
    i++;
}

// For-In Loop
console.log("For-In Loop:");
for (let index in numbers) {
    console.log(index);
}

// For-Of Loop
console.log("For-Of Loop:");
for (let value of numbers) {
    console.log(value);
}
```

## Question 8
**Write a function greet that takes a name and a callback function. The
greet function should call the callback with the name**

### Answer
**JavaScript Code**
```js
function greet(name, callback) {
    callback(name);
}

function display(name) {
    console.log("Hello " + name);
}

greet("Rahul", display);
```
**Output**

Hello Rahul

**Explanation**

- greet() accepts a name and a callback.
- display() acts as the callback.
- The callback is executed inside greet().

**Applications**

- Event handling
- API calls
- Timers
- Asynchronous programming

###Question 11:
**Given the array numbers = [1, 2, 3, 4, 5], perform the following: 1. Add 6 to the end.
2. Remove the first element.
3. Find the index of 4.
4. Check if all numbers are greater than 0.**
###Answer
**JavaScript Code**
```js
const numbers = [1, 2, 3, 4, 5];

// Add 6
numbers.push(6);

// Remove first element
numbers.shift();

// Find index of 4
console.log(numbers.indexOf(4));

// Check if all numbers are greater than 0
console.log(numbers.every(num => num > 0));

// Display updated array
console.log(numbers);
```

**Output**

2

true

[2, 3, 4, 5, 6]

**Explanation**

- push(6) adds 6 to the end.
- shift() removes the first element.
- indexOf(4) returns the index of 4.
- every() checks whether all elements satisfy the given condition.
