###Question 1 :
**Write code that creates a new paragraph element with the text "Welcome
to our website!" and adds it to a div with the id "content" on the webpage. Explain what createElement and appendChild methods do.**
###Answer
**HTML Code**

```html
<div id="content"></div>
```

**JavaScript Code**
```js
// Create a new paragraph element
const paragraph = document.createElement("p");

// Add text to the paragraph
paragraph.textContent = "Welcome to our website!";

// Find the div element
const content = document.getElementById("content");

// Add the paragraph to the div
content.appendChild(paragraph);
Output
Welcome to our website!

(The text appears inside the <div> on the webpage.)
```

**Explanation**

- document.createElement("p") creates a new <p> (paragraph) element.
- textContent sets the text inside the paragraph.
- document.getElementById("content") selects the <div> with the ID content.
- appendChild() adds the newly created paragraph as the last child of the selected <div>.

**What createElement() Does**

- Creates a new HTML element.
- Does not display it until it is added to the webpage.
- Used for dynamic content creation.

**What appendChild() Does**

- Adds a child element to an existing parent element.
- Inserts the new element into the webpage.
- Places the new element at the end of the parent's child list


###Question 2 :
**You need to change the appearance of a button when users hover over it.
Write code that finds a button with id "myButton" and changes its background color to
blue and text color to white using the element.style property.**
###Answer
**HTML Code**
```html
<button id="myButton">Hover Me</button>
```

**JavaScript Code**
```js
const button = document.getElementById("myButton");

button.addEventListener("mouseover", function () {
    button.style.backgroundColor = "blue";
    button.style.color = "white";
});

button.addEventListener("mouseout", function () {
    button.style.backgroundColor = "";
    button.style.color = "";
});
```

**Output**

- When the mouse pointer moves over the button:
- Background becomes blue.
- Text becomes white.
- When the mouse leaves the button:
- The button returns to its original appearance.

**Explanation**

- getElementById() selects the button.
- mouseover event occurs when the mouse enters the button.
- mouseout event occurs when the mouse leaves the button.
- style.backgroundColor changes the background color.
- style.color changes the text color.

###Question 4:
**Write a function named extractDomain that takes an email string like
"john@example.com" and returns the domain part ("example.com") using only
string methods. Do not use regular expressions.**
###Answer
**JavaScript Code**
```
function extractDomain(email) {
    const index = email.indexOf("@");
    return email.slice(index + 1);
}
```

console.log(extractDomain("john@example.com"));

**Output**

example.com

**Explanation**

- indexOf("@") finds the position of the @ symbol.
- slice(index + 1) extracts everything after the @.
- The function returns only the domain name.

###Question 5:
**Write a function called repeatMessage that takes a message string and
a number, and returns the message repeated that many times, separated by a space.
Do not use loops.
Example:
repeatMessage("hello", 3) should return "hello hello hello"**
###Answer
**Example:**

`repeatMessage("hello", 3)`

**Output:**

hello hello hello
Answer

**Introduction**

JavaScript provides the repeat() method to repeat a string multiple times. By combining it with trim(), the repeated string can be returned without extra spaces at the beginning or end.

**JavaScript Code**
```js
function repeatMessage(message, number) {
    return (message + " ").repeat(number).trim();
}

console.log(repeatMessage("hello", 3));
```
**Output**

hello hello hello

**Explanation**

- (message + " ") adds a space after each word.
- repeat(number) repeats the string the specified number of times.
- trim() removes the extra space at the end.

The Event Loop is a mechanism in JavaScript that allows asynchronous operations to work efficiently. JavaScript is single-threaded, meaning it executes one task at a time. Long-running tasks like timers or API calls are handled by the browser or Node.js, while the Event Loop checks when these tasks are complete and moves their callbacks to the Call Stack for execution.

**Given Code**

```js
console.log("Start");

setTimeout(function () {
    console.log("Timeout 1");
}, 0);

console.log("Middle");

setTimeout(function () {
    console.log("Timeout 2");
}, 0);

console.log("End");
```

- The second setTimeout() also schedules its callback.
- console.log("End") executes immediately.
- After the Call Stack becomes empty, the Event Loop moves the first callback into the Call Stack, printing "Timeout 1".

###Question 7:
**Write a JavaScript function named startCountdown that logs “Time left:
X” every second, starting from 5 down to 1, and finally logs “Time’s up!”. Use
setInterval and clearInterval only.**
###Answer
**JavaScript Code**
```js
function startCountdown() {
    let count = 5;

    const timer = setInterval(function () {
        console.log("Time left: " + count);

        count--;

        if (count === 0) {
            clearInterval(timer);
            console.log("Time's up!");
        }
    }, 1000);
}

startCountdown();
```

**Output**

Time left: 5

Time left: 4

Time left: 3

Time left: 2

Time left: 1

Time's up!

**Explanation**
- count starts from 5.
- Every second, setInterval() prints the remaining time.
- After printing 1, clearInterval() stops the timer.
- Finally, "Time's up!" is displayed.

###Question 8:
**Explain what Promises are in JavaScript and write code that creates a
Promise to simulate checking if a user is logged in. The Promise should resolve with
user data after 2 seconds if successful, or reject with an error message if the user is not
found.**
###Answer
A Promise is an object that represents the eventual completion or failure of an asynchronous operation. A Promise has three states:

Pending – Initial state.
Fulfilled (Resolved) – Operation completed successfully.
Rejected – Operation failed.

Promises simplify asynchronous programming and avoid deeply nested callback functions.

**JavaScript Code**
```js
function checkUser(isLoggedIn) {

    return new Promise((resolve, reject) => {

        setTimeout(() => {

            if (isLoggedIn) {
                resolve({
                    id: 1,
                    name: "Rahul",
                    email: "rahul@example.com"
                });
            } else {
                reject("Error: User not found");
            }

        }, 2000);

    });

}

checkUser(true)
    .then(user => {
        console.log("User Data:", user);
    })
    .catch(error => {
        console.log(error);
    });
```
**Output (If User Exists)**

User Data:
{
  id: 1,
  name: "Rahul",
  email: "rahul@example.com"
}

**Output (If User Does Not Exist)**

Error: User not found

###Question 9:
**Rewrite the following Promise-based code to use async/await syntax. The
code fetches user data and then fetches that user's posts. Include proper error handling.**
 ```js
function getUserDataAndPosts(userId) {
    return fetchUserData(userId)
        .then(userData => {
            return fetchUserPosts(userData.id)
                .then(posts => {
                    return { user: userData, posts: posts };
                });
        })
        .catch(error => {
            console.error('Error:', error);
            throw error;
        });
}
```
###Answer
**JavaScript Code**
```js
async function getUserDataAndPosts(userId) {

    try {

        const userData = await fetchUserData(userId);

        const posts = await fetchUserPosts(userData.id);

        return {
            user: userData,
            posts: posts
        };

    } catch (error) {

        console.error("Error:", error);
        throw error;

    }

}
```

**Explanation**

- async makes the function asynchronous.
- await pauses execution until the Promise is completed.
- try...catch handles errors.
- First, user data is fetched.
- Then, posts for that user are fetched.
- Finally, both are returned as a single object.

**Advantages of async/await**

- Cleaner code.
- Easier to understand.
- Avoids long .then() chains.
- Better error handling with try...catch.

**Write code using the fetch API to get weather data from a public API.
Display the weather information in a user-friendly format. Assume you have the
following configuration for the API.
const apiKey = 'abcd1234';  
const apiUrl =
`https://api.openweathermap.org/data/2.5/weather?q=Mumbai&appid=${apiK

###Answer
**JavaScript Code**
```js
const apiKey = "abcd1234";

const apiUrl =
`https://api.openweathermap.org/data/2.5/weather?q=Mumbai&appid=${apiKey}&units=metric`;

fetch(apiUrl)
    .then(response => response.json())
    .then(data => {

        console.log("Weather Information");
        console.log("-------------------");
        console.log("City:", data.name);
        console.log("Temperature:", data.main.temp + "°C");
        console.log("Weather:", data.weather[0].main);
        console.log("Humidity:", data.main.humidity + "%");
        console.log("Wind Speed:", data.wind.speed + " m/s");

    })
    .catch(error => {
        console.log("Error fetching weather data:", error);
    });
 ```
**Sample Output**

Weather Information

City: Mumbai

Temperature: 29°C

Weather: Clouds

Humidity: 78%

Wind Speed: 4.5 m/s

**Explanation**
- fetch() sends a request to the weather API.
- response.json() converts the response into a JavaScript object.
- Weather details such as city, temperature, humidity, weather condition, and - wind speed are displayed.
- .catch() handles any network or API errors.
