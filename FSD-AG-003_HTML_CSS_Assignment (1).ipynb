## Question 1
**Create a layout with three boxes side by side using Flexbox. The first box
should have twice the width of the other two. Provide the HTML and CSS code
snippets.**

### Answer
**HTML Code**
```html
<!DOCTYPE html>
<html>
<head>
    <title>Flexbox Layout</title>
    <link rel="stylesheet" href="style.css">
</head>
<body>

<div class="container">

    <div class="box box1">
        Box 1
    </div>

    <div class="box">
        Box 2
    </div>

    <div class="box">
        Box 3
    </div>

</div>

</body>
</html>
```
**CSS Code**
```css
body{
    margin:0;
    padding:20px;
    font-family:Arial, sans-serif;
    background:#f5f5f5;
}

.container{
    display:flex;
    gap:20px;
}

.box{
    flex:1;
    background:skyblue;
    padding:50px;
    text-align:center;
    font-size:24px;
    border-radius:10px;
}

.box1{
    flex:2;
    background:orange;
}
```
Explanation: Flexbox arranges items horizontally. flex:2 makes first box twice the width of the others (flex:1). gap adds spacing.

## Question 3
**Write a media query that changes the background color of the body to light grey when the screen width is less than 600px.**

### Answer
```css
body{
    background-color:white;
    font-family:Arial, sans-serif;
    margin:0;
    padding:20px;
}

@media screen and (max-width:600px){

    body{
        background-color:lightgrey;
    }

}
```
**Explanation**

**Normal Style**

body{
background:white;
}

This sets the default background color of the webpage to white.

**Media Query**

`@media screen and (max-width:600px)`

This means:
Apply the enclosed CSS rules only when the screen width is 600px or smaller.
Typically affects mobile devices.

**Background Change**

`body{
background-color:lightgrey;
}`

When the condition is satisfied, the background changes from white to light grey.

## Question 4
**Create a button that changes its background color to blue and scales up slightly when hovered over, using CSS transitions. Provide the HTML and CSS code snippets.**

### Answer
**HTML Code**
```html
<!DOCTYPE html>
<html>
<head>
    <title>Button Hover Effect</title>
    <link rel="stylesheet" href="style.css">
</head>
<body>

<button class="btn">Hover Me</button>

</body>
</html>
```

**CSS Code**

```css
body{
    font-family:Arial, sans-serif;
    padding:40px;
    background:#f5f5f5;
}

.btn{
    background:#555;
    color:white;
    border:none;
    padding:15px 30px;
    font-size:18px;
    border-radius:8px;
    cursor:pointer;
    transition: background-color 0.3s ease, transform 0.3s ease;
}

.btn:hover{
    background:blue;
    transform:scale(1.1);
}
```
Explanation: transition creates a smooth animation on hover.

Explanation: @keyframes continuously moves the square left to right.

## Question 7
**Difference**

### Answer
| Absolute Position                                    | Fixed Position                            |
| ---------------------------------------------------- | ----------------------------------------- |
| Positioned relative to the nearest positioned parent | Positioned relative to the browser window |
| Scrolls with the page                                | Stays fixed while scrolling               |
| Used inside containers                               | Used for headers, menus, chat buttons     |
| Changes position when parent moves                   | Always remains in the same place          |
| Not fixed on the screen                              | Fixed on the screen                       |

**Example of Absolute**

`
.box{
position:absolute;
top:50px;
left:100px;
}`

The box is placed relative to its parent element.

**Example of Fixed**

`
.box{
position:fixed;
bottom:20px;
right:20px;
}
`

The box remains at the bottom-right corner even while scrolling.

**position:absolute**
- Positioned relative to nearest positioned ancestor.
- Scrolls with page.
- Removed from normal document flow.

**position:fixed**
- Positioned relative to browser viewport.
- Remains visible during scrolling.
- Useful for sticky buttons, headers, navigation.

Conclusion: Absolute depends on parent; Fixed depends on viewport.

## Question 8
**Create a layered layout where a semi-transparent overlay covers an image using z-index. Provide the HTML and CSS code snippets.**

### Answer
**HTML Code**
```html
<!DOCTYPE html>
<html>
<head>
<title>Image Overlay</title>
<link rel="stylesheet" href="style.css">
</head>

<body>

<div class="container">

<img src="image.jpg" alt="Sample Image">

<div class="overlay">
Hello World
</div>

</div>

</body>
</html>
```
**CSS Code**
```css
.container{
position:relative;
width:400px;
}

.container img{
width:100%;
display:block;
}

.overlay{
position:absolute;
top:0;
left:0;
width:100%;
height:100%;
background:rgba(0,0,0,0.5);
color:white;
display:flex;
justify-content:center;
align-items:center;
font-size:24px;
z-index:2;
}
```
Explanation: z-index places the overlay above the image while rgba creates transparency.
