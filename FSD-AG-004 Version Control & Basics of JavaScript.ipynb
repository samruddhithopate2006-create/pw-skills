## Question 1
**Question 1: Explain the purpose of Version Control Systems like Git. Why are they essential in modern software development?.**

### Answer
**Introduction**

A Version Control System (VCS) is software that helps developers manage and track changes made to source code and project files over time. It stores every version of a project, allowing developers to return to previous versions whenever required.

Git is the most popular distributed Version Control System. It was developed by Linus Torvalds in 2005 for Linux kernel development. Today, Git is widely used by individuals, companies, and open-source communities.

Version control is an essential part of modern software development because software projects are usually developed by multiple developers working together. Git helps them collaborate efficiently without overwriting each other's work.

**Purpose of Version Control Systems**


- Track every change made to project files.
- Maintain a complete history of modifications.
- Allow multiple developers to work simultaneously.
- Recover deleted or modified files.
- Manage different versions of a project.
- Merge changes from different developers.
- Reduce conflicts during development.
- Improve project organization and teamwork.

**What is Git?**

Git is a Distributed Version Control System (DVCS).

In Git, every developer has a complete copy of the project repository on their computer.

This means developers can work even without an internet connection and synchronize their changes later.


**Features of Git**
- Distributed version control
- Fast performance
- Branching and merging
- Complete project history
- Open-source software

Why Git is Essential in Modern Software Development
1. Tracks Changes

Git records every modification made to the project.

Developers can easily identify:

Who made the change
When it was made
Why it was made
2. Collaboration

Many developers can work on the same project without affecting each other's work.

Example:

Developer A works on Login Page
Developer B works on Dashboard
Developer C works on Payment Module

Git later combines all these changes safely.

3. Backup

Every commit acts as a backup.

If files are accidentally deleted, previous versions can be restored.


4. Team Management

Managers can review code before merging.

This improves software quality.



## Question 2
**List the typical steps you would follow when working with Git to make changes to a project and share them on GitHub.**

### Answer
**Step 1: Create or Clone Repository**


If the project already exists on GitHub:

git clone https://github.com/username/project.git

This downloads the project to the local computer.


**Step 2: Check Repository Status**

git status

Shows modified, new, or deleted files.


**Step 3: Create or Switch Branch**

git checkout -b feature-login

Creates and switches to a new branch.


**Step 4: Make Changes**

Edit project files using any code editor such as VS Code.

**Step 5: Add Files**

git add .

Moves all modified files to the staging area.

**Step 6: Commit Changes**

git commit -m "Added login feature"

Creates a permanent snapshot of changes.

**Step 7: Push Changes**

git push origin feature-login

Uploads the branch to GitHub.

**Step 8: Create Pull Request**

Open GitHub.

Create a Pull Request.

Request code review.

Merge after approval.

## Question 3
**What is the purpose of a .gitignore file in a Git repository? Provide an
example of how to ignore all .log files.**

### Answer
**Introduction**

A .gitignore file tells Git which files or folders should not be tracked or uploaded to the repository.

Many projects contain temporary files, log files, cache files, compiled files, and sensitive information that should not be shared.

The .gitignore file prevents these unnecessary files from being committed.

**Purpose of .gitignore**
Ignore temporary files.
Ignore log files.
Ignore compiled files.
Ignore cache folders.
Ignore sensitive files.


**Example**

To ignore all log files:

*.log


**Other Examples**

Ignore node_modules folder

- node_modules/

Ignore environment file

- .env

Ignore cache files

- .cache/

Ignore all text files

-*.txt

**Advantages**
- Prevents unnecessary uploads.
- Improves security.
- Keeps repository organized.
- Reduces storage space.

## Question 6
**Declare a constant named PI with the value 3.14159. Attempting to
reassign it should result in an error. Demonstrate this behavior by writing out the outcome.**

### Answer
The const keyword is used to declare constants in JavaScript. Once a value is assigned to a constant, it cannot be changed.

**Code**

const PI = 3.14159;

console.log(PI);

PI = 3.14;

**Output**

3.14159

TypeError: Assignment to constant variable.

**Explanation**

const PI = 3.14159; creates a constant named PI.

Attempting to assign a new value (PI = 3.14) causes an error because constants cannot be reassigned.

## Question 7
**You are given the following array:const colors = ["red", "green", "blue", "yellow", "purple"];1. Access and print the first element of the array.
2. Access and print the last element of the array.
3. Access and print the third element of the array.**

### Answer
An array stores multiple values in a single variable. Array indexing starts from 0, meaning the first element is at index 0.

Code

```js

const colors = ["red", "green", "blue", "yellow", "purple"];

console.log(colors[0]);

console.log(colors[colors.length - 1]);

console.log(colors[2]);
```

**Output**

red
purple
blue

**Explanation**

colors[0] → First element (red)

colors[colors.length - 1] → Last element (purple)

colors[2] → Third element (blue).

###Question 9:
**Identify the data types of the following values: 1. "Hello"     2. 42     3. true     4. undefined**
###Answer

- "Hello"	 : String

- 42	  : Number

- true	: Boolean

- undefined	 : Undefined

**Explanation**
String – Represents text enclosed in quotes.

Number – Represents numeric values.

Boolean – Represents either true or false.

Undefined – A variable that has been declared but has not been assigned a value.

**Example Code**

console.log(typeof "Hello");      // string
console.log(typeof 42);           // number
console.log(typeof true);         // boolean
console.log(typeof undefined);    // undefined

**Output**

string

number

boolean

undefined
