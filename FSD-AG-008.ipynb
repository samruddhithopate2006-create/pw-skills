###Question 1 :
**What is React?**
###Answer:
React is a free and open-source JavaScript library used to build fast, interactive, and dynamic user interfaces (UI) for web applications. It was developed by Facebook (Meta) in 2013 and is widely used for creating Single Page Applications (SPAs).

React follows a component-based architecture, where the user interface is divided into small, reusable components. Each component has its own logic and can be reused throughout the application.

One of React's biggest advantages is the Virtual DOM, which updates only the changed parts of a webpage instead of reloading the entire page. This makes React applications faster and more efficient.

**Features of React**

- Component-based architecture
- Virtual DOM
- JSX (JavaScript XML)
- One-way data flow
- Reusable components
- High performance
- Easy to learn
- Strong community support

**Advantages**
- Faster rendering
- Reusable code
- Better performance
- Easy maintenance
- Large ecosystem

**Applications**
- E-commerce websites
- Social media applications
- Dashboards
- Online learning platforms
- Banking websites
- Portfolio websites

###Question 2:
**Write a React functional component named Greeting that accepts a name prop and displays: "Hello, [name]!".**
###Answer
A functional component is a JavaScript function that returns JSX. Props (Properties) are used to pass data from a parent component to a child component.

**Greeting Component**

###Question 4:
**Explain how props support React's one-way data flow.**
###Answer
Props (Properties) are used to pass data from a parent component to a child component. React follows a one-way data flow, meaning information always moves from parent to child.

The child component can read and display the data but cannot directly modify it. If data needs to change, the parent component updates it and passes the new value again.

**Example**
```react
function Child(props) {
    return <h2>{props.message}</h2>;
}

function App() {
    return (
        <Child message="Hello React" />
    );
}

export default App;
```

**Advantages**
- Easy to understand.
- Better code organization.
- Simplifies debugging.
- Prevents accidental data modification.
- Makes applications predictable.

###Question 5:
**Create a functional component ClickMeButton that renders a button. It accepts onClick and label props. When clicked, it executes the passed onClick handler.**
###Answer
React allows event handling through props. A parent component can pass an event handler function and button text to a child component, making the component reusable.
```react
ClickMeButton Component
function ClickMeButton(props) {
    return (
        <button onClick={props.onClick}>
            {props.label}
        </button>
    );
}

export default ClickMeButton;
```
**Using the Component**
```react
import ClickMeButton from "./ClickMeButton";

function App() {

    function handleClick() {
        alert("Button Clicked!");
    }

    return (
        <ClickMeButton
            label="Click Me"
            onClick={handleClick}
        />
    );
}

export default App;
```
**Output**

Initially, the page displays:

Click Me

After clicking the button:

Alert: Button Clicked!

**Explanation**
- The label prop sets the button text.
- The onClick prop receives the click handler function.
- When the button is clicked, the function passed from the parent component is executed.
- This demonstrates event handling and one-way data flow in React.

###Question 6:
**Given const fruits = ['Apple', 'Banana', 'Cherry'], write a component FruitList that renders an unordered list of these fruits using .map() and includes a key.**
###Answer
In React, the map() method is used to display multiple elements from an array. Each rendered element should have a unique key so React can efficiently update the UI.

**React Code**
```react
function FruitList() {

    const fruits = ["Apple", "Banana", "Cherry"];

    return (
        <ul>
            {fruits.map((fruit, index) => (
                <li key={index}>{fruit}</li>
            ))}
        </ul>
    );
}

export default FruitList;
```
**Output**

• Apple

• Banana

• Cherry

**Explanation**

- map() loops through each fruit in the array.
- Each fruit is displayed inside an <li> element.
- The key prop helps React identify each list item.
- React updates only the changed items, improving performance

###Question 7:
**Why is the key attribute important when rendering lists in React? What goes wrong if keys are missing or array index is misused?**
###Answer
The key attribute is a special React prop used to uniquely identify each item in a list. It helps React determine which items have changed, been added, or removed during updates.

**Why is key Important?**

- Identifies each list item uniquely.
- Helps React update only changed items.
- Improves rendering performance.
- Prevents unnecessary re-rendering.
- Makes list updates more efficient.
- Problems if Keys are Missing
- React displays warning messages.
- UI updates become less efficient.
- Components may lose their state.
- Incorrect items may be updated or reused.
- Problems with Using Array Index as Key

**Using the array index as a key is acceptable only for static lists. If items are added, removed, or reordered:**

- Wrong items may be updated.
- State may shift to another item.
- Performance decreases.
- Unexpected UI behaviour can occur.

**Example**
```
const fruits = ["Apple", "Banana", "Cherry"];

function FruitList() {
    return (
        <ul>
            {fruits.map((fruit, index) => (
                <li key={index}>{fruit}</li>
            ))}
        </ul>
    );
}
```

###Question 8:
**Write a component UserList that receives a users prop—an array of objects each with id and name. Render each user's name in a <p> tag using map() and key.**
###Answer
Props allow data to be passed from a parent component to a child component. The child component can use map() to display each user.

**UserList Component**

```react
function UserList({ users }) {

    return (
        <div>
            {users.map(user => (
                <p key={user.id}>
                    {user.name}
                </p>
            ))}
        </div>
    );
}

export default UserList;
```
**App Component**
```react
import UserList from "./UserList";

function App() {

    const users = [
        { id: 1, name: "Rahul" },
        { id: 2, name: "Priya" },
        { id: 3, name: "Amit" }
    ];

    return <UserList users={users} />;
}

export default App;
```
**Output**

Rahul

Priya

Amit

**Explanation**

- users is received as a prop.
- map() loops through the array.
- Each user's name is displayed inside a <p> element.
- user.id is used as the key.

###Question 9:
**Create an App component that holds an array of numbers, passes them to a NumberList child component for rendering, and passes a click handler function to log the clicked number. Show one-way data flow and event handling.**
###Answer
React follows one-way data flow, where data moves from the parent component to the child component through props. Event handlers can also be passed as props.

**NumberList Component**
```react
function NumberList({ numbers, onNumberClick }) {

    return (
        <div>
            {numbers.map(number => (
                <button
                    key={number}
                    onClick={() => onNumberClick(number)}
                >
                    {number}
                </button>
            ))}
        </div>
    );
}

export default NumberList;
```
**App Component**
```react
import NumberList from "./NumberList";

function App() {

    const numbers = [10, 20, 30, 40];

    function handleNumberClick(number) {
        console.log("Clicked:", number);
    }

    return (
        <NumberList
            numbers={numbers}
            onNumberClick={handleNumberClick}
        />
    );
}

export default App;
```
**Output**

10   20   30   40

Clicking 20 prints:

Clicked: 20

**Explanation**

- numbers are stored in the parent component.
- The child component receives them through props.
- Clicking a button calls the parent's click handler.
- This demonstrates one-way data flow and event handling.

###Question 10:
**Design a component ProductList that receives products (array of objects with id, name, price). Render a list of product names and prices using .map(). Each item when clicked should call a prop onProductSelect(product). Include the proper key.**
###Answer
This example demonstrates rendering a list of objects using map(), assigning a unique key, and handling click events through props.

**ProductList Component**
```react
function ProductList({ products, onProductSelect }) {

    return (
        <div>
            {products.map(product => (
                <div
                    key={product.id}
                    onClick={() => onProductSelect(product)}
                >
                    <h3>{product.name}</h3>
                    <p>Price: ₹{product.price}</p>
                </div>
            ))}
        </div>
    );
}

export default ProductList;
```
**App Component**
```react
import ProductList from "./ProductList";

function App() {

    const products = [
        { id: 1, name: "Laptop", price: 50000 },
        { id: 2, name: "Mouse", price: 800 },
        { id: 3, name: "Keyboard", price: 1500 }
    ];

    function handleProductSelect(product) {
        console.log(product);
    }

    return (
        <ProductList
            products={products}
            onProductSelect={handleProductSelect}
        />
    );
}

export default App;
```
**Output**

Laptop
Price: ₹50000

Mouse
Price: ₹800

Keyboard
Price: ₹1500

Clicking Laptop prints:

{ id: 1, name: "Laptop", price: 50000 }

**Explanation**

- products are passed as props from the parent component.
- map() renders each product.
- product.id is used as the unique key.
- Clicking a product calls onProductSelect(product) with the selected product.
