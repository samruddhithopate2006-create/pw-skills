##Question 1: Create a minimal router with two pages: Home (/) and About (/about). Show the links to move between pages.
##Answer:

React Router is used to handle navigation between different pages or components in a React application without completely reloading the webpage.

First, install React Router:

*npm install react-router-dom*

**App.jsx**
```
import { BrowserRouter, Routes, Route, Link } from "react-router-dom";


function Home() {
  return <h2>Home Page</h2>;
}


function About() {
  return <h2>About Page</h2>;
}


function App() {
  return (
    <BrowserRouter>
      <nav>
        <Link to="/">Home</Link> |{" "}
        <Link to="/about">About</Link>
      </nav>


      <Routes>
        <Route path="/" element={<Home />} />
        <Route path="/about" element={<About />} />
      </Routes>
    </BrowserRouter>
  );
}


export default App;
```
**Explanation:**
- BrowserRouter enables routing in the React application.
- Routes contains all the application routes.
- Route connects a URL path with a React component.
- Link creates navigation links without refreshing the entire page.
- / displays the Home component.
- /about displays the About component.

**Navigation:**
- Home → /

 - About → /about

When the user clicks Home or About, React Router changes the displayed component without performing a full browser page reload.

##Question 2: After a user clicks “Save”, programmatically go to /thank-you.
##Answer:

React Router provides the useNavigate() hook for programmatic navigation. It is useful when navigation should happen as the result of an action, such as submitting a form or clicking a button.
```
import { BrowserRouter, Routes, Route, useNavigate } from "react-router-dom";


function SavePage() {
  const navigate = useNavigate();


  const handleSave = () => {
    console.log("Data saved");
    navigate("/thank-you");
  };


  return (
    <button onClick={handleSave}>
      Save
    </button>
  );
}


function ThankYou() {
  return <h2>Thank You!</h2>;
}


function App() {
  return (
    <BrowserRouter>
      <Routes>
        <Route path="/" element={<SavePage />} />
        <Route path="/thank-you" element={<ThankYou />} />
      </Routes>
    </BrowserRouter>
  );
}


export default App;
```
**Explanation:**

- const navigate = useNavigate();

- creates a navigation function.

- When the Save button is clicked:

- navigate("/thank-you");

- React Router changes the URL to: /thank-you  
and displays the component associated with that route.

**Flow:**

User clicks Save

handleSave()

Data is saved

navigate("/thank-you")

Thank You page

This is called programmatic navigation because the application changes the route using JavaScript instead of the user clicking a <Link>.

##Question 3: Build a DashboardLayout that shows a heading and a menu. Inside it, render DashboardHome at /dashboard and Settings at /dashboard/settings.
##Answer:

A layout route allows multiple related pages to share a common layout, such as a heading, navigation menu, sidebar, or header.

React Router's Outlet is used to display the child route inside the layout.
```
import {
  BrowserRouter,
  Routes,
  Route,
  Link,
  Outlet
} from "react-router-dom";


function DashboardLayout() {
  return (
    <div>
      <h1>Dashboard</h1>


      <nav>
        <Link to="/dashboard">Home</Link> |{" "}
        <Link to="/dashboard/settings">Settings</Link>
      </nav>


      <hr />


      <Outlet />
    </div>
  );
}


function DashboardHome() {
  return <h2>Dashboard Home</h2>;
}


function Settings() {
  return <h2>Settings</h2>;
}


function App() {
  return (
    <BrowserRouter>
      <Routes>
        <Route path="/dashboard" element={<DashboardLayout />}>
          <Route index element={<DashboardHome />} />
          <Route path="settings" element={<Settings />} />
        </Route>
      </Routes>
    </BrowserRouter>
  );
}


export default App;
```
**Explanation:**

- *The parent route is:*

<Route path="/dashboard" element={<DashboardLayout />}>

- *The child route:*

<Route index element={<DashboardHome />} />

- *renders DashboardHome at:*

/dashboard

- *The second child:*

<Route path="settings" element={<Settings />} />

- *renders Settings at:*

/dashboard/settings

The <Outlet /> determines where the child component appears inside DashboardLayout.



##Question 4: Create a route /users/:userId. In the UserPage component, read the userId from the URL and display it.
##Answer:

React Router provides the useParams() hook to access dynamic parameters from a URL.

**Code:**
```
import {
  BrowserRouter,
  Routes,
  Route,
  useParams
} from "react-router-dom";


function UserPage() {
  const { userId } = useParams();


  return (
    <div>
      <h2>User Page</h2>
      <p>User ID: {userId}</p>
    </div>
  );
}


function App() {
  return (
    <BrowserRouter>
      <Routes>
        <Route path="/users/:userId" element={<UserPage />} />
      </Routes>
    </BrowserRouter>
  );
}


export default App;
```
*Example URL:*

http://localhost:5173/users/101

*The route:* /users/:userId

contains :userId as a dynamic route parameter.

Therefore: const { userId } = useParams();

will give: 101

**Output:**

User Page


User ID: 101

If the user visits:

/users/500

the output becomes:

User ID: 500

**Simple Flow:**

/users/101

:userId = 101

useParams()

userId = "101"

Display User ID

This is useful for pages such as user profiles, product details, blog posts, and order details.



##Question 5: Only logged-in users should see /dashboard. Others should be redirected to /login. Write a minimal ProtectedRoute wrapper.
##Answer:

A Protected Route is a route that can only be accessed when a particular condition is satisfied.

For example, a dashboard should be accessible only to logged-in users.

React Router provides the Navigate component to redirect users to another route.

**Code:**
```
import {
  BrowserRouter,
  Routes,
  Route,
  Navigate
} from "react-router-dom";


function ProtectedRoute({ isLoggedIn, children }) {
  if (!isLoggedIn) {
    return <Navigate to="/login" replace />;
  }


  return children;
}


function Dashboard() {
  return <h2>Dashboard</h2>;
}


function Login() {
  return <h2>Login Page</h2>;
}


function App() {
  const isLoggedIn = false;


  return (
    <BrowserRouter>
      <Routes>
        <Route
          path="/dashboard"
          element={
            <ProtectedRoute isLoggedIn={isLoggedIn}>
              <Dashboard />
            </ProtectedRoute>
          }
        />


        <Route path="/login" element={<Login />} />
      </Routes>
    </BrowserRouter>
  );
}


export default App;
```
**Explanation:**

- The ProtectedRoute receives: isLoggedIn

If the user is not logged in:

if (!isLoggedIn) {
  return <Navigate to="/login" replace />;
}

the user is redirected to:

/login

- If the user is logged in:

return children;

the requested component, such as Dashboard, is displayed.


##Question 6: Create a counter slice with increment and decrement. Configure the store and provide it to React.
##Answer:

Redux Toolkit simplifies Redux development by providing functions for creating slices and configuring the Redux store.

*First, install the required packages:*

npm install @reduxjs/toolkit react-redux

**Step 1: Create the counter slice**

*Create:*

src/features/counter/counterSlice.js
```
import { createSlice } from "@reduxjs/toolkit";


const counterSlice = createSlice({
  name: "counter",


  initialState: {
    value: 0
  },


  reducers: {
    increment: (state) => {
      state.value += 1;
    },


    decrement: (state) => {
      state.value -= 1;
    }
  }
});


export const { increment, decrement } = counterSlice.actions;


export default counterSlice.reducer;
```
**Step 2: Configure the Redux store**

*Create:*

src/store.js
```
import { configureStore } from "@reduxjs/toolkit";
import counterReducer from "./features/counter/counterSlice";


export const store = configureStore({
  reducer: {
    counter: counterReducer
  }
});
```
Step 3: Provide the store to React

In main.jsx:
```
import React from "react";
import ReactDOM from "react-dom/client";
import { Provider } from "react-redux";


import App from "./App";
import { store } from "./store";


ReactDOM.createRoot(document.getElementById("root")).render(
  <React.StrictMode>
    <Provider store={store}>
      <App />
    </Provider>
  </React.StrictMode>
);
```
**Explanation:**

createSlice()

*Creates:*

State
Reducer functions
Actions

In this example, the initial state is:
```
{
  value: 0
}
```
The increment reducer increases the value:

state.value += 1;

The decrement reducer decreases it:
```
state.value -= 1;
configureStore()

Creates the Redux store and registers the counter reducer.

Provider
<Provider store={store}>
  <App />
</Provider>
```
makes the Redux store available to React components throughout the application.

##Question 7: Build a Counter component that shows the value and buttons to increment and decrement using Redux hooks.
##Answer:

*React Redux provides hooks such as:*

- useSelector() → reads data from the Redux store.
- useDispatch() → sends actions to the Redux store.
Counter.jsx
```
import { useSelector, useDispatch } from "react-redux";
import {
  increment,
  decrement
} from "./features/counter/counterSlice";


function Counter() {
  const count = useSelector((state) => state.counter.value);


  const dispatch = useDispatch();


  return (
    <div>
      <h2>Counter: {count}</h2>


      <button onClick={() => dispatch(increment())}>
        +1
      </button>


      <button onClick={() => dispatch(decrement())}>
        -1
      </button>
    </div>
  );
}


export default Counter;
```
**Explanation:**

The following code reads the counter value:
```
const count = useSelector(
  (state) => state.counter.value
);
```

Counter: 0

dispatch(increment())

the Redux increment action is dispatched.

The state changes:

0 → 1 → 2 → 3

When the user clicks:

dispatch(decrement())

the value decreases:

3 → 2 → 1

**Simple Flow:**

User clicks +1

dispatch(increment())

Redux Reducer

Counter state increases

##Question 8: Create a fetchProducts thunk that loads from /api/products. Handle loading and error in the slice.
##Answer:

A thunk is used when Redux needs to perform an asynchronous operation, such as fetching data from an API.

Redux Toolkit provides createAsyncThunk() for this purpose.

**productSlice.js**
```
import {
  createSlice,
  createAsyncThunk
} from "@reduxjs/toolkit";


// Async thunk
export const fetchProducts = createAsyncThunk(
  "products/fetchProducts",


  async () => {
    const response = await fetch("/api/products");


    if (!response.ok) {
      throw new Error("Failed to fetch products");
    }


    return await response.json();
  }
);


const productSlice = createSlice({
  name: "products",


  initialState: {
    items: [],
    loading: false,
    error: null
  },


  reducers: {},


  extraReducers: (builder) => {
    builder


      // Request started
      .addCase(fetchProducts.pending, (state) => {
        state.loading = true;
        state.error = null;
      })


      // Request successful
      .addCase(fetchProducts.fulfilled, (state, action) => {
        state.loading = false;
        state.items = action.payload;
      })


      // Request failed
      .addCase(fetchProducts.rejected, (state, action) => {
        state.loading = false;
        state.error = action.error.message;
      });
  }
});


export default productSlice.reducer;
```
**Configure the Store**
```
import { configureStore } from "@reduxjs/toolkit";
import productReducer from "./productSlice";


export const store = configureStore({
  reducer: {
    products: productReducer
  }
});
Using the Thunk in a Component
import { useEffect } from "react";
import { useDispatch, useSelector } from "react-redux";
import { fetchProducts } from "./productSlice";


function Products() {
  const dispatch = useDispatch();


  const {
    items,
    loading,
    error
  } = useSelector((state) => state.products);


  useEffect(() => {
    dispatch(fetchProducts());
  }, [dispatch]);


  if (loading) {
    return <p>Loading products...</p>;
  }


  if (error) {
    return <p>Error: {error}</p>;
  }


  return (
    <div>
      <h2>Products</h2>


      {items.map((product) => (
        <p key={product.id}>
          {product.name}
        </p>
      ))}
    </div>
  );
  ```
**Explanation:**

When:

dispatch(fetchProducts());

is called, three possible states are handled.

1. Pending

While the request is running:

loading = true

The UI can display:

Loading products...

2. Fulfilled

If the request succeeds:

loading = false
items = products

The products are displayed.

3. Rejected

If the request fails:

loading = false
error = error message

The UI displays the error.




##Question 9: Style a “Buy” button using Tailwind. Use a blue background, white text, and change the background shade on hover and focus.
##Answer:

Tailwind CSS provides utility classes that can be directly added to HTML or JSX elements.

A simple Buy button can be created as follows:
```

function ProductList() {
  const products = [
    { id: 1, name: "Laptop", price: 50000 },
    { id: 2, name: "Phone", price: 25000 },
    { id: 3, name: "Headphones", price: 3000 },
    { id: 4, name: "Keyboard", price: 1500 },
    { id: 5, name: "Mouse", price: 800 },
    { id: 6, name: "Monitor", price: 15000 }
  ];


  return (
    <div className="grid grid-cols-1 md:grid-cols-3 gap-6">
      {products.map((product) => (
        <div
          key={product.id}
          className="product-card"
        >
          <h2>{product.name}</h2>
          <p>₹{product.price}</p>
        </div>
      ))}
    </div>
  );
}


export default ProductList;
```
Reusing Utility Classes with @apply

In the Tailwind CSS stylesheet, we can create a reusable class:
```
@tailwind base;
@tailwind components;
@tailwind utilities;


.product-card {
  @apply p-5 bg-white rounded-lg shadow;
  @apply transition duration-300;
  @apply hover:shadow-lg hover:-translate-y-1;
}
```

Now every element with:

className="product-card"

gets all those styles.

**Explanation of the Responsive Grid:**

className="grid grid-cols-1 md:grid-cols-3 gap-6"

means:

grid → Uses CSS Grid.
grid-cols-1 → One column by default.
md:grid-cols-3 → Three columns from the medium breakpoint and above.
gap-6 → Adds spacing between products.
Layout:

Small screen:
```
┌───────────────┐
│    Laptop     │
├───────────────┤
│     Phone     │
├───────────────┤
│  Headphones   │
├───────────────┤
│   Keyboard    │
└───────────────┘

Medium/Large screen:

┌─────────┐ ┌─────────┐ ┌─────────┐
│ Laptop  │ │  Phone  │ │Headphone│
└─────────┘ └─────────┘ └─────────┘


┌─────────┐ ┌─────────┐ ┌─────────┐
│Keyboard │ │  Mouse  │ │ Monitor │
└─────────┘ └─────────┘ └─────────┘
```

**Hover Transition:**

*These classes:*
```
@apply transition duration-300;
@apply hover:shadow-lg hover:-translate-y-1;
```
make the product card smoothly change when the mouse moves over it.

transition → Enables smooth animation.

duration-300 → Makes the transition approximately 300 ms.

hover:shadow-lg → Increases the shadow on hover.

hover:-translate-y-1 → Moves the card slightly upward.

Why use @apply?

*Instead of repeatedly writing many Tailwind classes:*

className="p-5 bg-white rounded-lg shadow transition duration-300

 hover:shadow-lg hover:-translate-y-1"

*we can create:*
```
.product-card {
  @apply p-5 bg-white rounded-lg shadow transition duration-300 hover:shadow-lg hover:-translate-y-1;
}
```
*Then simply use:*

className="product-card"

This makes repeated styling easier to maintain and reuse.
