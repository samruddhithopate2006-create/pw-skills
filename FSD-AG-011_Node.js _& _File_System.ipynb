##Question 5: Using the fs/promises module, write a script to asynchronously read the content of a file named data.txt and log it to the console.
##Answer:

Node.js provides the File System (fs) module for working with files. The fs/promises API provides Promise-based methods that can be used with async/await.

**Step 1: Create data.txt**

Create a file named:

data.txt

Put some content inside it, for example:

Hello, this is data from the file.

Learning Node.js File System module.

**Step 2: Create the JavaScript file**

Create:

readFile.js
**Step 3: Write the code**
```
const fs = require("fs/promises");
async function readData() {
    try {
        const data = await fs.readFile("data.txt", "utf8");


        console.log(data);
    } catch (error) {
        console.log("Error reading file:", error.message);
    }
}

readData();
```
**Step 4: Execute the program**

node readFile.js

*Output:*

Learning Node.js File System module.

- const fs = require("fs/promises");

Imports the Promise-based File System module.

- fs.readFile("data.txt", "utf8")

asynchronously reads the contents of data.txt.

The "utf8" parameter tells Node.js to return the file contents as a string rather than a Buffer.

- await

waits for the file-reading operation to complete.

- console.log(data);

displays the contents of the file.

**Why use fs/promises?**

The Promise-based API works naturally with async/await and makes asynchronous file operations easier to read.

Error Handling

##Question 6: Using the fs/promises module, write a script to create a new file named log.txt and write the string "Task completed." into it.
##Answer:

The fs/promises module provides asynchronous file-system operations using Promises. The writeFile() method can be used to create a new file and write content into it.
```
const fs = require("fs/promises");


async function createLogFile() {
    try {
        await fs.writeFile("log.txt", "Task completed.");


        console.log("File created successfully.");
    } catch (error) {
        console.log("Error:", error.message);
    }
}


createLogFile();
```
**Output:**

File created successfully.

After executing the program, a new file named log.txt will be created.

*Content of log.txt:*

Task completed.

**Explanation:**
- require("fs/promises") imports the Promise-based File System module.
- writeFile() creates the file if it does not already exist.
- If log.txt already exists, its existing content is replaced.
- await waits for the asynchronous operation to finish.
- try...catch handles possible errors.

*Execute:*

node createLog.js

##Question 8: Create a basic HTTP server using the http module that listens on port 3000 and responds to all requests with "Server is running."
##Answer:

Node.js has a built-in http module that can be used to create an HTTP server without installing an external package.

**Code:**
```
const http = require("http");

const server = http.createServer((req, res) => {


    res.writeHead(200, {
        "Content-Type": "text/plain"
    });


    res.end("Server is running.");
});


server.listen(3000, () => {
    console.log("Server running on port 3000");
});
```
*Run the program:*

Save it as: server.js

Then execute: node server.js

Terminal Output:  Server running on port 3000

Open the following URL in a browser:  http://localhost:3000

**Browser Output:**

Server is running.

**Explanation:**

- require("http") imports Node.js's built-in HTTP module.
- http.createServer() creates an HTTP server.
- req contains information about the client's request.
- res is used to send a response.
- res.writeHead(200) sends the HTTP success status.
- Content-Type: text/plain indicates that the response is plain text.
- res.end() sends the response and finishes the request.
- server.listen(3000) makes the server listen on port 3000.

**Request-Response Flow:**

Browser

GET http://localhost:3000

Node.js HTTP Server

res.end("Server is running.")

##Question 9: Enhance the previous HTTP server to handle two different routes: / should respond with "Homepage", and /about should respond with "About Us".
##Answer:

The req.url property can be used to determine which URL the client has requested. We can use conditions to send different responses for different routes.

**Code:**
```
const http = require("http");


const server = http.createServer((req, res) => {


    res.writeHead(200, {
        "Content-Type": "text/plain"
    });


    if (req.url === "/") {
        res.end("Homepage");
    }
    else if (req.url === "/about") {
        res.end("About Us");
    }
    else {
        res.writeHead(404, {
            "Content-Type": "text/plain"
        });


        res.end("Page Not Found");
    }
});


server.listen(3000, () => {
    console.log("Server running on port 3000");
});
```

**Test the Home Route:**

Open: http://localhost:3000/

**Output:**

Homepage
Test the About Route:

Open: http://localhost:3000/about

**Output:**

About Us
Invalid Route:

For example:

http://localhost:3000/contact

**Output:**

Page Not Found
Explanation:
req.url

contains the URL path requested by the client.

For: http://localhost:3000/

the value is:/

For: http://localhost:3000/about

the value is:/about

The server checks these values and sends the appropriate response.

**Route Table:**

URL	Response

/about	About Us

##Question 10: Write a script that efficiently reads a large file (large.txt) and outputs its content to the console using streams.
##Answer:

When working with a very large file, reading the entire file at once using readFile() may require a large amount of memory.

Streams allow Node.js to read the file in smaller chunks. This makes them suitable for large files.

**Code:**
```
const fs = require("fs");


const readStream = fs.createReadStream("large.txt", {
    encoding: "utf8"
});


readStream.on("data", (chunk) => {
    console.log(chunk);
});


readStream.on("end", () => {
    console.log("File reading completed.");
});


readStream.on("error", (error) => {
    console.log("Error reading file:", error.message);
});
```
**Explanation:**

fs.createReadStream("large.txt")

creates a readable stream for the file.

Instead of loading the complete file into memory, Node.js reads it chunk by chunk.
```
data Event
readStream.on("data", (chunk) => {
    console.log(chunk);
});
```
The data event is triggered whenever a new piece of the file is available.

The chunk contains a portion of the file.

end Event
```
readStream.on("end", () => {
    console.log("File reading completed.");
});
```
The end event occurs when the complete file has been read.
error Event
```
readStream.on("error", (error) => {
    console.log("Error reading file:", error.message);
});
```
This handles problems such as the file not existing.

**Execute:**

*Save the code as:*

readLargeFile.js

*Make sure large.txt is in the same folder:*

project/

│

├── large.txt

└── readLargeFile.js

*Run:*

node readLargeFile.js

**Simple Working:**

large.txt

createReadStream()

Read small chunk

"data" event

Display chunk

Read next chunk

"data" event

...

"end" event

File reading completed.
