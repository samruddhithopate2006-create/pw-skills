## Question 1: Custom Event Emitter with `userLoggedIn` Event

### Answer:

Node.js provides an **EventEmitter** class through its built-in `events` module. It allows us to create and handle custom events.

In this program, we create a custom event called `userLoggedIn`. When this event occurs, it sends the username as data, also called a **payload**, to the event listener. The listener then prints a welcome message for that user.

### Code:

```js
const EventEmitter = require('events');

// Create an EventEmitter object
const eventEmitter = new EventEmitter();

// Create a listener for the userLoggedIn event
eventEmitter.on('userLoggedIn', (username) => {
    console.log(`Welcome, ${username}! You have successfully logged in.`);
});

// Emit or fire the event with a username payload
eventEmitter.emit('userLoggedIn', 'Samruddhi');
```

### Output:

```text
Welcome, Samruddhi! You have successfully logged in.
```

### Explanation:

1. `require('events')`

   * The `events` module is a built-in Node.js module.
   * It provides the `EventEmitter` class.
   * We do not need to install it using npm.

2. `new EventEmitter()`

   * An object of the `EventEmitter` class is created.
   * This object can create and manage custom events.

```js
const eventEmitter = new EventEmitter();
```

3. `eventEmitter.on()`

   * The `.on()` method is used to register or attach an event listener.
   * The listener waits for the `userLoggedIn` event to occur.

```js
eventEmitter.on('userLoggedIn', (username) => {
    console.log(`Welcome, ${username}!`);
});
```

4. `eventEmitter.emit()`

   * The `.emit()` method is used to fire or trigger an event.
   * Here, the event name is `userLoggedIn`.
   * The value `'Samruddhi'` is passed as the payload.

```js
eventEmitter.emit('userLoggedIn', 'Samruddhi');
```

5. The value passed through `emit()` is received by the listener:

```js
(username) => {
    console.log(`Welcome, ${username}!`);
}
```

### Answer:
### Code:

```js
const crypto = require('crypto');

// Original string
const password = 'password123';

// Create SHA256 hash
const hash = crypto
    .createHash('sha256')
    .update(password)
    .digest('hex');

// Print the hash
console.log('Original Password:', password);
console.log('SHA256 Hash:', hash);
```

### Example Output:

```text
Original Password: password123
SHA256 Hash:
ef92b778bafe771e89245b89ecbc08a44a4fce7e...
```

The complete SHA256 hash is a hexadecimal value containing 64 characters.

## Question 4: Write a Node.js script that reads a `PORT` environment variable. If it's not set, it should default to `3000`.

### Answer:

Environment variables are external values that can be provided to a Node.js application. They are commonly used to store configuration values such as:

* Port number
* Database URL
* Username
* Password
* API keys
* Application environment

In Node.js, environment variables can be accessed using:

```js
process.env
```

The following program checks whether a `PORT` environment variable exists. If it exists, the application uses that port. Otherwise, it uses port `3000` as the default value.

### Code:

```js
const PORT = process.env.PORT || 3000;

console.log(`Server will run on port: ${PORT}`);
```

### Output When `PORT` Is Not Set:

```text
Server will run on port: 3000
```

### Output When `PORT` Is Set:

For example, if:

```text
PORT=5000
```

then the output will be:

```text
Server will run on port: 5000
```

### Explanation:

#### 1. `process`

`process` is a global object provided by Node.js. It contains information about the currently running Node.js process.

#### 2. `process.env`

The `env` property contains environment variables.

```js
process.env.PORT
```

This reads the value of the `PORT` environment variable.

---

#### 3. Default value using `||`

```js
const PORT = process.env.PORT || 3000;
```

The logical OR operator `||` works as follows:

```text
If process.env.PORT has a value
        ↓
Use process.env.PORT

Otherwise
        ↓
Use 3000
```

Therefore:

```js
const PORT = process.env.PORT || 3000;
```

means:

> Use the PORT environment variable if it is available; otherwise, use port 3000.

### Example with an HTTP Server:

```js
const http = require('http');

const PORT = process.env.PORT || 3000;

const server = http.createServer((req, res) => {
    res.writeHead(200, {
        'Content-Type': 'text/plain'
    });

    res.end('Server is running successfully!');
});

server.listen(PORT, () => {
    console.log(`Server is running on port ${PORT}`);
});
```

## Question 5: Write the minimal code for an Express.js server that listens on port 3000 and responds to GET requests at the root URL (`/`) with `"Hello, Express!"`.

### Answer:
### Minimal Express Server Code:

```js
const express = require('express');

const app = express();

app.get('/', (req, res) => {
    res.send('Hello, Express!');
});

app.listen(3000, () => {
    console.log('Server is running on http://localhost:3000');
});
```

### Answer:
### SQL Code:

```sql
CREATE TABLE users (
    id INT AUTO_INCREMENT PRIMARY KEY,
    email VARCHAR(255) NOT NULL UNIQUE,
    password_hash VARCHAR(255) NOT NULL,
    created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP
);
```

### Explanation:

#### 1. `CREATE TABLE users`

```sql
CREATE TABLE users
```

This command creates a new table called `users`.

---

#### 2. `id INT AUTO_INCREMENT PRIMARY KEY`

```sql
id INT AUTO_INCREMENT PRIMARY KEY
```

This creates the `id` column.

* `INT` stores integer values.
* `AUTO_INCREMENT` automatically generates the next ID for every new user.
* `PRIMARY KEY` uniquely identifies each record.

For example:

| id | email                                       |
| -- | ------------------------------------------- |
| 1  | [test@example.com](mailto:test@example.com) |
| 2  | [user@gmail.com](mailto:user@gmail.com)     |
| 3  | [admin@gmail.com](mailto:admin@gmail.com)   |

The ID is automatically generated when a new user is inserted.

---

#### 3. `email VARCHAR(255) NOT NULL UNIQUE`

```sql
email VARCHAR(255) NOT NULL UNIQUE
```

This column stores the user's email address.

* `VARCHAR(255)` stores text up to 255 characters.
* `NOT NULL` means the email cannot be empty.
* `UNIQUE` prevents two users from having the same email address.

---

#### 4. `password_hash VARCHAR(255) NOT NULL`

```sql
password_hash VARCHAR(255) NOT NULL
```

This column stores the **hashed version of the user's password**.

In real applications, passwords should not be stored as plain text.

For example:

```text
Incorrect:
password = password123
```

Instead:

```text
Correct:
password_hash = encrypted/hashed value
```

## Question 10: Given a `users` table and a `posts` table (with a `user_id` foreign key), write a query to retrieve all posts and the email of the user who wrote them.

### Answer:

When data is stored in multiple related tables, we can use a SQL `JOIN` operation to combine data from those tables.

In this example:

### `users` Table:

| id | email                                       |
| -- | ------------------------------------------- |
| 1  | [test@example.com](mailto:test@example.com) |
| 2  | [user@gmail.com](mailto:user@gmail.com)     |

### `posts` Table:

| id | title         | content          | user_id |
| -- | ------------- | ---------------- | ------- |
| 1  | My First Post | Hello World      | 1       |
| 2  | Node.js       | Learning Node.js | 2       |
| 3  | SQL JOIN      | Learning SQL     | 1       |

The `user_id` column in the `posts` table refers to the `id` column in the `users` table.

Therefore, we can connect the two tables using a `JOIN`.

### SQL Code:

```sql
SELECT posts.*, users.email
FROM posts
JOIN users
ON posts.user_id = users.id;
```

### Explanation:

#### 1. `SELECT posts.*, users.email`

```sql
SELECT posts.*, users.email
```

This selects:

* All columns from the `posts` table.
* The `email` column from the `users` table.

`posts.*` means all columns from the `posts` table.

---

#### 2. `FROM posts`

```sql
FROM posts
```

This means the query starts with the `posts` table.

---

#### 3. `JOIN users`

```sql
JOIN users
```

This joins the `users` table with the `posts` table.

---

#### 4. Join condition

```sql
ON posts.user_id = users.id;
```

This connects each post with the user who created it.

The relationship is:

```text
posts.user_id
       │
       │ matches
       ▼
users.id
``
### Using Table Aliases:

The same query can also be written in a shorter and cleaner form using aliases:

```sql

SELECT p.*, u.email
FROM posts AS p
JOIN users AS u
ON p.user_id = u.id;

```

Here:

```text
p = posts
u = users
```

This is commonly used when queries become larger and contain multiple tables.
