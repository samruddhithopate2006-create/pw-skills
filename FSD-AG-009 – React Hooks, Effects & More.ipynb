###Question 1
**In simple words, what are React Hooks? Create a small component that
shows a button. When the button is clicked, toggle the text between “On” and “Off”.**
###Answer:
**What are React Hooks?**

React Hooks are special functions introduced in React 16.8 that allow functional components to use React features such as state, lifecycle methods, context, and references without writing class components.

Hooks make React code simpler, cleaner, and easier to understand. They also help developers reuse logic between different components.

**Some commonly used React Hooks are:**

- useState() – Manages state.
- useEffect() – Performs side effects like API calls or timers.
- useContext() – Shares data without prop drilling.
- useReducer() – Manages complex state.
- useRef() – Accesses DOM elements directly.

**React Code**
```jsx
import { useState } from "react";

function ToggleButton() {

    const [status, setStatus] = useState("Off");

    function toggleStatus() {
        if (status === "Off") {
            setStatus("On");
        } else {
            setStatus("Off");
        }
    }

    return (
        <button onClick={toggleStatus}>
            {status}
        </button>
    );
}

export default ToggleButton;
```
**Output**

Initially:

Off

After clicking:

On

After clicking again:

Off

**Explanation**
- useState("Off") creates a state variable.
- Clicking the button changes the value between On and Off.
- React automatically updates the UI whenever the state changes.

###Question 2 :
**Make a counter with a button. Whenever the count changes, log “Count
changed” using useEffect.**
###Answer:  
**About useEffect**

The useEffect() Hook is used to perform side effects after a component renders. Examples include API calls, timers, logging messages, and updating the page title.

**React Code**
```react

import { useState, useEffect } from "react";

function Counter() {

    const [count, setCount] = useState(0);

    useEffect(() => {
        console.log("Count changed");
    }, [count]);

    return (
        <div>
            <h2>{count}</h2>

            <button onClick={() => setCount(count + 1)}>
                Increase
            </button>
        </div>
    );
}

export default Counter;
```

**Output**

0
[Increase]

After clicking once:

1

Console:

Count changed

**Explanation**
- count stores the counter value.
- Clicking the button increases the value.
- useEffect() runs whenever count changes.
- The message "Count changed" appears in the browser console.

###Question 3 :
**Write a component that *fetches* users from https://jsonplaceholder.typicode.com/users only once on mount and shows how many users were loaded.**
###Answer:
**About the Program**

React uses useEffect() to fetch data from APIs. By passing an empty dependency array ([]), the API is called only once when the component loads.

**React Code**

import { useState, useEffect } from "react";

function UserCount() {

    const [users, setUsers] = useState([]);

    useEffect(() => {

        fetch("https://jsonplaceholder.typicode.com/users")
            .then(response => response.json())
            .then(data => {
                setUsers(data);
            });

    }, []);

    return (
        <div>
            <h2>Total Users: {users.length}</h2>
        </div>
    );
}

export default UserCount;

**Output**

Total Users: 10

**Explanation**
- fetch() requests data from the API.
- The response is converted into JSON.
- setUsers() stores the user data.
- users.length displays the total number of users.
- The API runs only once because of the empty dependency array.

###Question 4 :
**Create a component that starts a timer with setInterval to increase a “seconds” counter every second. Stop the timer when the component unmounts.**
###Answer:
The setInterval() function repeatedly executes code after a specified interval. In React, useEffect() is used to start the timer, while the cleanup function stops it using clearInterval().

**React Code**
```react
import { useState, useEffect } from "react";

function Timer() {

    const [seconds, setSeconds] = useState(0);

    useEffect(() => {

        const timer = setInterval(() => {
            setSeconds(prev => prev + 1);
        }, 1000);

        return () => {
            clearInterval(timer);
        };

    }, []);

    return (
        <h2>Seconds: {seconds}</h2>
    );
}

export default Timer;
```
**Output**

- Seconds: 0
- Seconds: 1
- Seconds: 2
- Seconds: 3
...

**Explanation**

- setInterval() runs every second.
- setSeconds() updates the counter.
- clearInterval() stops the timer when the component is removed.
- This prevents unnecessary background execution and memory leaks.

###Question 5 :
**Build a small form with a text input. If the input is empty, show “Please type your name”. If it has text, show “Hello, <name>”.**
###Answer:
This example demonstrates a controlled component, where the input value is stored in React state using useState(). As the user types, React updates the displayed message automatically.

**React Code**
```react
import { useState } from "react";

function NameForm() {

    const [name, setName] = useState("");

    return (
        <div>

            <input
                type="text"
                placeholder="Enter your name"
                value={name}
                onChange={(e) => setName(e.target.value)}
            />

            {name === "" ? (
                <p>Please type your name</p>
            ) : (
                <p>Hello, {name}</p>
            )}

        </div>
    );
}

export default NameForm;
```
**Output**

Initially:

Please type your name

If the user enters Rahul:

Hello, Rahul

**Explanation**
- useState("") stores the input value.
- onChange() updates the state whenever the user types.
- If the input is empty, the message "Please type your name" is displayed.
- Otherwise, React displays "Hello, <name>" using the entered value.

###Question 6
**You have a theme value (“light” or “dark”) that is needed by a deep child button. Show how to pass it without sending theme through every parent prop.**
###Answer
Passing data through many intermediate components is called prop drilling. React provides the Context API and the useContext() Hook to solve this problem. Using Context, a value can be shared directly with any child component without passing it through every parent component.

```react
{/*Step 1: Create Context*/}
import { createContext } from "react";

export const ThemeContext = createContext();

{/*Step 2: App Component*/}

import { ThemeContext } from "./ThemeContext";
import Child from "./Child";

function App() {

    const theme = "dark";

    return (
        <ThemeContext.Provider value={theme}>
            <Child />
        </ThemeContext.Provider>
    );
}

export default App;

{/*Step 3: Child Component*/}

import { useContext } from "react";
import { ThemeContext } from "./ThemeContext";

function Child() {

    const theme = useContext(ThemeContext);

    return (
        <button>
            Current Theme: {theme}
        </button>
    );
}

export default Child;
```

**Output**

Current Theme: dark

**Explanation**
- createContext() creates a Context object.
- ThemeContext.Provider shares the theme value.
- useContext() accesses the value directly.
- No need to pass theme through multiple components.

###Question 7 :
**Make a counter with two buttons: “+1” and “-1”. Use useReducer instead of useState.**
###Answer:
useReducer() is a React Hook used for managing complex state. Instead of directly updating state, actions are dispatched to a reducer function, which returns the new state.

**React Code**
```react
import { useReducer } from "react";

function reducer(state, action) {

    switch (action.type) {

        case "increment":
            return state + 1;

        case "decrement":
            return state - 1;

        default:
            return state;
    }
}

function Counter() {

    const [count, dispatch] = useReducer(reducer, 0);

    return (
        <div>

            <h2>{count}</h2>

            <button onClick={() => dispatch({ type: "increment" })}>
                +1
            </button>

            <button onClick={() => dispatch({ type: "decrement" })}>
                -1
            </button>

        </div>
    );
}

export default Counter;
```

**Output**

Initially:

0

After clicking +1:

1

After clicking -1:

0

**Explanation**
- useReducer() initializes the counter with 0.
- Clicking +1 dispatches the "increment" action.
- Clicking -1 dispatches the "decrement" action.
- The reducer updates the state based on the action type.

###Question 8 :
**Given an array nums = [1,2,3,4] and a filter value min, show how to compute a filtered list only when nums or min change. Also pass a stable click handler to a child list item.**
###Answer:
**React Code**

```react
import { useMemo, useCallback } from "react";

function NumberList({ min }) {

    const nums = [1, 2, 3, 4];

    const filteredNumbers = useMemo(() => {
        return nums.filter(num => num >= min);
    }, [nums, min]);

    const handleClick = useCallback((number) => {
        console.log("Clicked:", number);
    }, []);

    return (
        <div>
            {filteredNumbers.map(number => (
                <p
                    key={number}
                    onClick={() => handleClick(number)}
                >
                    {number}
                </p>
            ))}
        </div>
    );
}

export default NumberList;
```

**Output (If min = 3)**

3
4

Clicking 3 prints:

Clicked: 3

**Explanation**

- useMemo() recalculates the filtered list only when nums or min changes.
- useCallback() keeps the click function stable between renders.
- This improves application performance.

###Question 9 :
**Create a controlled input for email. As the user types, show “Email: <value>”. Why do we keep the input’s value in state?**
###Answer:
**React Code**
```react
import { useState } from "react";

function EmailInput() {

    const [email, setEmail] = useState("");

    return (
        <div>

            <input
                type="email"
                placeholder="Enter your email"
                value={email}
                onChange={(e) => setEmail(e.target.value)}
            />

            <p>Email: {email}</p>

        </div>
    );
}

export default EmailInput;
```
**Output**

Initially:

Email:

After typing:

Email: abc@gmail.com

**Why do we keep the input's value in state?**
- Keeps the UI synchronized with user input.
- Makes validation easier.
- Enables real-time updates.
- Simplifies form submission.
- React becomes the single source of truth for the input value.

###Question 10 :
**Make a button that focuses an input using a ref. Return adjacent elements without extra wrapper nodes. In one sentence, say what React’s update process does.**
###Answer:
**React Code**
```react
import { useRef } from "react";

function FocusInput() {

    const inputRef = useRef();

    function handleFocus() {
        inputRef.current.focus();
    }

    return (
        <>
            <input
                ref={inputRef}
                type="text"
                placeholder="Enter your name"
            />

            <button onClick={handleFocus}>
                Focus Input
            </button>
        </>
    );
}

export default FocusInput;
```

**Output**

- Initially:

[ Enter your name ]

[ Focus Input ]

- After clicking Focus Input, the cursor automatically moves to the input field.

- React's Update Process (One Sentence)

React compares the Virtual DOM with the previous Virtual DOM and updates only the changed parts of the Real DOM, making the application faster and more efficient.

**Explanation**

- useRef() creates a reference to the input element.
- ref={inputRef} connects the reference to the input.
- inputRef.current.focus() places the cursor inside the input.
- React Fragment (<>...</>) returns multiple elements without creating unnecessary wrapper nodes.
