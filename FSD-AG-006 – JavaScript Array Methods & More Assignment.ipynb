###Question 1
**Explain what the forEach method does in JavaScript arrays and write code
that uses forEach to display each student's name from an array of student objects to the
console.
// Array of student objects
const students = [
{ name: "Rahul", age: 20 },
{ name: "Priya", age: 19 },
{ name: "Amit", age: 21 },
{ name: "Sneha", age: 20 }
];**
###Answer
The forEach() method is one of the most commonly used array methods in JavaScript. It is used to execute a function once for every element in an array.

Unlike the map() method, forEach() does not create a new array. It simply performs an operation such as displaying data, updating values, or calling another function for each element.

**The syntax of forEach() is:**

```js
array.forEach(function(element){
    // code
});
```

**It is mainly used for:**

- Displaying array elements
- Printing values
- Updating objects
- Performing calculations
- Calling functions repeatedly
- JavaScript Code

// Array of student objects

```js
const students = [
    { name: "Rahul", age: 20 },
    { name: "Priya", age: 19 },
    { name: "Amit", age: 21 },
    { name: "Sneha", age: 20 }
];

// Display student names
students.forEach(function(student) {
    console.log(student.name);
});
```
**Output**

Rahul

Priya

Amit

Sneha

**Explanation**
- students.forEach() visits every object in the array.
- student represents the current object.
- student.name accesses the name property.
- console.log() prints each student's name.

**Advantages of forEach()**
- Easy to read.
- Removes the need for traditional loops.
- Executes code for every array element.
- Useful for printing and updating values.
- Improves code readability.

**Applications**
- Displaying records
- Printing database results
- Processing user lists
- Showing product names
- Creating menus

###Question 2
 **You have an array of product prices in rupees. Write code using the map method to create a new array that contains the same prices but with 18% GST added to each price.const originalPrices =[100, 250, 500, 750, 1000];**
###Answer
The map() method creates a new array by applying a function to each element of an existing array.

Unlike forEach(), map() returns a completely new array without changing the original array.

**It is widely used for:**

- Price calculations
- Data conversion
- Creating new arrays
- Mathematical operations

**JavaScript Code**

```js
const originalPrices = [100, 250, 500, 750, 1000];

const gstPrices = originalPrices.map(function(price) {
    return price + (price * 18 / 100);
});

console.log("Original Prices:", originalPrices);
console.log("Prices with GST:", gstPrices);
```

**Output**

Original Prices:

[100, 250, 500, 750, 1000]

Prices with GST:

[118, 295, 590, 885, 1180]

**Explanation**
- map() visits every price.
- Calculates 18% GST.
- Returns the updated price.
- Stores the result in a new array.
- Original array remains unchanged.

**Advantages of map()**
- Creates a new array.
- Original array is safe.
- Less code.

**Applications**
- GST calculation
- Currency conversion
- Student marks conversion

###Question 5
**Explain how the sort method works in JavaScript and write code to sort an
array of employee objects by their salary in descending order (highest salary first).
// Array of employee objects
const employees = [
{ name: "Rajesh", salary: 45000 },
{ name: "Priya", salary: 55000 },
{ name: "Amit", salary: 40000 },
{ name: "Kavya", salary: 60000 }
];**
###Answer
The sort() method arranges the elements of an array in a specified order.

For numbers and objects, a comparison function is used to define the sorting order.

Ascending → smallest to largest

Descending → largest to smallest

**JavaScript Code**
```js
const employees = [
{ name: "Rajesh", salary: 45000 },
{ name: "Priya", salary: 55000 },
{ name: "Amit", salary: 40000 },
{ name: "Kavya", salary: 60000 }
];

employees.sort((a, b) => b.salary - a.salary);

console.log(employees);

//Output

[
 { name: "Kavya", salary: 60000 },
 { name: "Priya", salary: 55000 },
 { name: "Rajesh", salary: 45000 },
 { name: "Amit", salary: 40000 }
]
```
**Explanation**

The comparison function

b.salary - a.salary

sorts salaries from highest to lowest.

If we used

a.salary - b.salary

the salaries would be sorted in ascending order.

**Advantages of sort()**

- Organizes data.
- Supports ascending and descending order.
- Works with numbers, strings, and objects.
- Easy to use.
- Improves data presentation.

**Applications**

- Salary sorting
- Product price sorting
- Student marks ranking
- Alphabetical lists
- Leaderboards

###Question 6
**Write code using the reduce method to calculate the total value of all items
in a shopping cart. Each item has a price and quantity. Also find the most expensive
single item in the cart.
// Shopping cart with items
const cart = [
{ name: "Laptop", price: 50000, quantity: 1 },
{ name: "Mouse", price: 500, quantity: 2 },
{ name: "Keyboard", price: 1500, quantity: 1 },
{ name: "Monitor", price: 15000, quantity: 1 }
];**
###Answer
**JavaScript Code**
```js
// Shopping cart with items
const cart = [
    { name: "Laptop", price: 50000, quantity: 1 },
    { name: "Mouse", price: 500, quantity: 2 },
    { name: "Keyboard", price: 1500, quantity: 1 },
    { name: "Monitor", price: 15000, quantity: 1 }
];

// Calculate total value
const totalValue = cart.reduce((total, item) => {
    return total + (item.price * item.quantity);
}, 0);

// Find most expensive item
const expensiveItem = cart.reduce((max, item) => {
    return item.price > max.price ? item : max;
});

console.log("Total Value:", totalValue);
console.log("Most Expensive Item:", expensiveItem);
```

**Output**

Total Value: 67500

Most Expensive Item:
{
  name: "Laptop",
  price: 50000,
  quantity: 1
}

**Explanation**

- reduce() calculates the total cost by multiplying price × quantity for each item.
- Another reduce() compares item prices to find the most expensive item.

// Student object

###Question 9
**Write a function called calculateTotal that can accept any number of
prices as arguments and returns their sum. Use the rest operator to handle the variable number of parameters.**
###Answer
The rest operator (...) allows a function to accept any number of arguments. These arguments are stored as an array inside the function.

**JavaScript Code**
```js
function calculateTotal(...prices) {

    return prices.reduce((total, price) => total + price, 0);

}

console.log(calculateTotal(100, 200, 300));
console.log(calculateTotal(50, 150, 250, 400));
```

**Output**

600

850

**Explanation**

- ...prices collects all arguments into an array.
- reduce() adds all prices together.
- The function works with any number of values.

**Advantages**
- Accepts unlimited arguments.
- Simplifies function parameters.
- Works well with arrays.
- Makes code flexible.

###Question 10
**You have two arrays of student names and need to create a new array
that contains all students from both arrays. Also, you have two objects representing a student's basic info and marks. Write code using the spread operator to solve both scenarios.
const classA = ["Rahul", "Priya", "Amit"];  
const classB = ["Sneha", "Vikram", "Pooja"];**
###Answer

**Introduction**

The spread operator (...) expands the elements of an array or object. It is commonly used to merge arrays and objects.

**JavaScript Code**
```js
const classA = ["Rahul", "Priya", "Amit"];
const classB = ["Sneha", "Vikram", "Pooja"];

// Merge arrays
const allStudents = [...classA, ...classB];

console.log(allStudents);

// Merge objects
const basicInfo = {
    name: "Rahul",
    age: 20
};

const marks = {
    math: 90,
    science: 85
};

const student = {
    ...basicInfo,
    ...marks
};

console.log(student);
```

**Output**

[
  "Rahul",
  "Priya",
  "Amit",
  "Sneha",
  "Vikram",
  "Pooja"
]

{
  name: "Rahul",
  age: 20,
  math: 90,
  science: 85
}

**Explanation**

...classA and ...classB combine two arrays into one.
...basicInfo and ...marks merge two objects into a single object.
